# **What is LLMOps**

Systematic set of practices, tools and processes to follow while building and managing the complete lifecycle of a LLM-based application in production  

Steps:  
- Choose a foundation model
- Build the application 
- Design and manage your prompts (Prompt Versioning, Experiment Tracking)
- Evaluate application
- Add Guardrails and Safety
- Deploy, Monitor and Improve (API Endpoints, Containerization, CI/CD, Post-deployment Monitoring and Observability)

### Tools  

- Orchestration : LangGraph  
- Prompt Management : LangFuse 
- Experiment Tracking : MLFlow  
- Evaluation : DeepEval / RAGAS
- Observability : LangFuse / LangSmith  
- Guardrails : Guardrails AI
- Serving : FastAPI / Streamlit
- Containerization : Docker
- CICD : GitHub Actions / DeepEval CLI / Docker Hub
- Infrastructure : AWS
- Infra Management : Prometheus (Tracking), Grafana (Dashboards)

### MLOps VS LLMOps  

| MLOps | LLMOps |
|-------|--------|
| For Machine Learning apps | For LLM based apps|
| No model available so train it | Models already available so no need to train|
| Deterministic outputs so easier to <br>manage input output | Probabilistic outputs. Difficult to manage input output |
| Total cost static. <br>Same cost for same amount of computation | Total cost dynamic based on input and output tokens and pricing |
| Constant number of steps in workflow and execution | Variable steps execution of workflow | 
| Well managed data is extremely important | Well managed prompts are extremely important |
| Evaluation less difficult | Evaluation more difficult as unknown output |
| No guardrails needed | Guardrails needed |
| Drift in data to be tracked | Drift in tokens and inputs to be tracked |
